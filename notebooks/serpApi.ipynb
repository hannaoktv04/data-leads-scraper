```python
import requests
import pandas as pd
import time
import os
from dotenv import load_dotenv

load_dotenv()

API_KEY = os.getenv("SERPAPI_API_KEY")

if not API_KEY:
    raise ValueError("SERPAPI_API_KEY belum dikonfigurasi.")

KEYWORDS = [
    "Cafe Terbesar",
    "Cafe Luas Jakarta",
    "Cafe Kapasitas Besar",
    "Cafe untuk Gathering",
    "Cafe Event Space",
    "Cafe Outdoor Luas",
    "Coffee Roastery",
    "Roastery and Cafe",
    "Compound Space",
    "Communal Space Cafe",
    "Industrial Coffee Shop",
    "Creative Hub Cafe",
    "Mega Cafe",
    "Spacious Coffee Shop",
    "Spacious Cafe",
    "Large Capacity Cafe",
    "Premium Cafe & Resto",
    "Bistro & Cafe",
    "Artisan Coffee Roaster",
    "Cafe Group",
    "F&B Group",
    "Cafe Franchise"
]

AREAS = [
    "Jakarta Timur",
    "Rawamangun",
    "Duren Sawit",
    "Cibubur",
    "Pondok Kelapa",
    "Matraman",
    "Jatinegara",
    "Cawang",
    "Buaran"

]

OUTPUT = "jakarta.xlsx"

MAX_PAGE_PER_QUERY = 50
MAX_TOTAL_QUERY = 400

all_data = []
query_count = 0

for area in AREAS:
    for keyword in KEYWORDS:
        if query_count >= MAX_TOTAL_QUERY:
            break

        query = f"{keyword} {area}"
        query_count += 1

        print(f"\n[{query_count}] Mencari: {query}")

        for page in range(MAX_PAGE_PER_QUERY):
            start = page * 100

            params = {
                "engine": "google_maps",
                "q": query,
                "type": "search",
                "start": start,
                "api_key": API_KEY
            }

            try:
                res = requests.get(
                    "https://serpapi.com/search.json",
                    params=params,
                    timeout=30
                )
                data = res.json()
            except Exception as e:
                print("Request gagal:", e)
                break

            if "error" in data:
                print("Error:", data["error"])
                break

            results = data.get("local_results", [])
            print(f"  Halaman {page + 1}: {len(results)} data")

            if not results:
                break

            for place in results:
                gps = place.get("gps_coordinates", {})

                all_data.append({
                    "Query": query,
                    "Nama": place.get("title"),
                    "Kategori": place.get("type"),
                    "Alamat": place.get("address"),
                    "Rating": place.get("rating"),
                    "Jumlah Review": place.get("reviews"),
                    "Telepon": place.get("phone"),
                    "Website": place.get("website"),
                    "Link Google Maps": place.get("link"),
                    "Latitude": gps.get("latitude"),
                    "Longitude": gps.get("longitude"),
                    "Data ID": place.get("data_id")
                })

            time.sleep(1)

    if query_count >= MAX_TOTAL_QUERY:
        break

df = pd.DataFrame(all_data)

if df.empty:
    print("Data kosong.")
else:
    print("\nTotal sebelum hapus duplikat:", len(df))

    df = df.drop_duplicates(subset=["Nama", "Alamat", "Telepon"], keep="first")

    print("Total setelah hapus duplikat:", len(df))

    df.to_excel(OUTPUT, index=False)
    files.download(OUTPUT)

df.head()
```
